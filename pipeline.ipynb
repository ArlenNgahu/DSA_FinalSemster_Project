Cell 1: Setup and Load Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', 100)
pd.set_option('display.width', 200)

# Upload the file in Colab first (or mount Drive)
# from google.colab import files
# uploaded = files.upload()

df = pd.read_csv('C:\\Users\\HP\\Documents\\DSA3900\\DSA_FinalSemster_Project\\data\\ken2015.csv', low_memory=False)

print("Shape of dataset:", df.shape)
print("\nFirst 5 rows:")
df.head()
```

Cell 2: Basic Inspection

```python

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\n--- Rural / Urban ---")
print(df['urbrur'].value_counts(dropna=False))
# 1 = Urban, 2 = Rural

print("\n--- Sex ---")
print(df['sex'].value_counts(dropna=False))

print("\n--- Age ---")
print(df['age'].describe())
```

Cell 3: Clean Coded Missing Values

```python

# Common coded missing values in STEPS surveys: 888, 999, 77, 0 (for some measurements)

# Blood pressure
bp_cols = ['m4a', 'm4b', 'm5a', 'm5b', 'm6a', 'm6b']
for col in bp_cols:
    if col in df.columns:
        df[col] = df[col].replace([0, 888, 999], np.nan)

# Height and Weight
for col in ['m11', 'm12']:
    if col in df.columns:
        df[col] = df[col].replace([0, 888, 888.8, 999], np.nan)

# Glucose and Cholesterol
for col in ['b5', 'b8', 'b9']:
    if col in df.columns:
        df[col] = df[col].replace([0, 777, 888, 999], np.nan)

print("Coded missing values cleaned.")
print("\nMissing values after cleaning (key variables):")
print(df[['m4a','m4b','m5a','m5b','m11','m12','b5','b8']].isna().sum())
```

Cell 4: Create Average BP and BMI

```python
# Average of the two (or three) blood pressure readings
df['systolic'] = df[['m4a', 'm5a']].mean(axis=1)
df['diastolic'] = df[['m4b', 'm5b']].mean(axis=1)

# BMI = weight (kg) / height (m)^2
df['height_m'] = df['m11'] / 100
df['bmi'] = df['m12'] / (df['height_m'] ** 2)

print("Average Systolic BP:")
print(df['systolic'].describe())

print("\nBMI:")
print(df['bmi'].describe())
```

Cell 5: Create Binary CVD Risk Target

```python
# We define High Risk (1) if ANY of the following is true:
# 1. Hypertension: Systolic ≥ 140 OR Diastolic ≥ 90
# 2. Obesity: BMI ≥ 30
# 3. Raised blood glucose: ≥ 7.0 mmol/L (diabetes range)
# 4. Current tobacco user (t1 == 1)

df['hypertension'] = ((df['systolic'] >= 140) | (df['diastolic'] >= 90)).astype(int)
df['obesity'] = (df['bmi'] >= 30).astype(int)
df['raised_glucose'] = (df['b5'] >= 7.0).astype(int)
df['current_smoker'] = (df['t1'] == 1).astype(int)

# Final binary target
df['cvd_high_risk'] = (
    (df['hypertension'] == 1) |
    (df['obesity'] == 1) |
    (df['raised_glucose'] == 1) |
    (df['current_smoker'] == 1)
).astype(int)

print("Target variable distribution:")
print(df['cvd_high_risk'].value_counts(dropna=False))
print("\nPercentage High Risk:")
print(df['cvd_high_risk'].value_counts(normalize=True) * 100)
```

Cell 6: Select Modelling Features

```python

# Create a clean modelling dataset
model_df = df[[
    'age', 'sex', 'urbrur',
    'systolic', 'diastolic', 'bmi',
    'b5',          # blood glucose
    'b8',          # total cholesterol
    't1',          # tobacco
    'a1',          # alcohol
    'cvd_high_risk'
]].copy()

# Rename for clarity
model_df = model_df.rename(columns={
    'urbrur': 'rural_urban',
    'b5': 'glucose',
    'b8': 'cholesterol',
    't1': 'tobacco',
    'a1': 'alcohol'
})

print("Modelling dataset shape:", model_df.shape)
print("\nMissing values in modelling features:")
print(model_df.isna().sum())
model_df.head()
```

Cell 7: Final Cleaning for Modelling

```python
# Encode sex
model_df['sex'] = model_df['sex'].map({'Men': 1, 'Women': 0})

# Encode rural_urban (1=Urban, 2=Rural → make Rural=1 for easier interpretation)
model_df['rural'] = (model_df['rural_urban'] == 2).astype(int)

# Drop original rural_urban
model_df = model_df.drop(columns=['rural_urban'])

# Drop rows with missing target or critical features
model_df = model_df.dropna(subset=['cvd_high_risk', 'age', 'sex', 'systolic', 'diastolic', 'bmi'])

print("Final modelling dataset shape after cleaning:", model_df.shape)
print("\nTarget balance:")
print(model_df['cvd_high_risk'].value_counts(normalize=True) * 100)
print("\nRural vs Urban:")
print(model_df['rural'].value_counts())
```

Cell 8: Prepare Features and Target

```python

# Features (X) and Target (y)
feature_cols = ['age', 'sex', 'rural', 'systolic', 'diastolic', 'bmi', 
                'glucose', 'cholesterol', 'tobacco', 'alcohol']

X = model_df[feature_cols].copy()
y = model_df['cvd_high_risk'].copy()

# Fill remaining missing values with median (simple and robust)
X = X.fillna(X.median(numeric_only=True))

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nAny remaining missing values?", X.isna().sum().sum())
X.head()
```

Cell 9: Stratified Train-Test Split

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set size:", X_train.shape[0])
print("Test set size:", X_test.shape[0])
print("\nTrain target distribution:")
print(y_train.value_counts(normalize=True).round(3) * 100)
print("\nTest target distribution:")
print(y_test.value_counts(normalize=True).round(3) * 100)
```

Cell 10: Train Models

```python


from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, 
                             f1_score, roc_auc_score, classification_report, 
                             confusion_matrix)

# Random Forest
rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    min_samples_split=10,
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train, y_train)

# XGBoost
xgb = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    eval_metric='logloss',
    use_label_encoder=False
)
xgb.fit(X_train, y_train)

print("Both models trained successfully.")
```

Cell 11: Model Evaluation

```python
def evaluate_model(model, X_test, y_test, model_name):
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:, 1]
    
    print(f"\n===== {model_name} Results =====")
    print(f"Accuracy : {accuracy_score(y_test, y_pred):.4f}")
    print(f"Precision: {precision_score(y_test, y_pred):.4f}")
    print(f"Recall   : {recall_score(y_test, y_pred):.4f}")
    print(f"F1-Score : {f1_score(y_test, y_pred):.4f}")
    print(f"AUC      : {roc_auc_score(y_test, y_proba):.4f}")
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))

evaluate_model(rf, X_test, y_test, "Random Forest")
evaluate_model(xgb, X_test, y_test, "XGBoost")
```


Cell 12: Corrected Target (Hypertension) - No Leakage


```python


# Target: Hypertension (clinically meaningful)
model_df['hypertension'] = ((model_df['systolic'] >= 140) | (model_df['diastolic'] >= 90)).astype(int)

print("Hypertension distribution:")
print(model_df['hypertension'].value_counts(normalize=True).round(3) * 100)

# Features that do NOT include systolic/diastolic
feature_cols_clean = ['age', 'sex', 'rural', 'bmi', 'glucose', 'cholesterol', 'tobacco', 'alcohol']

X = model_df[feature_cols_clean].copy()
y = model_df['hypertension'].copy()

# Fill missing values
X = X.fillna(X.median(numeric_only=True))

print("\nClean features shape:", X.shape)
print("Any missing values left?", X.isna().sum().sum())
```


Cell 13: New Stratified Split


```python


from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train size:", X_train.shape[0])
print("Test size :", X_test.shape[0])
print("\nTrain hypertension rate:", y_train.mean().round(3))
print("Test hypertension rate :", y_test.mean().round(3))
```


Cell 14: Retrain Random Forest & XGBoost


```python


from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

rf = RandomForestClassifier(
    n_estimators=300,
    max_depth=8,
    min_samples_split=15,
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train, y_train)

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.08,
    random_state=42,
    eval_metric='logloss'
)
xgb.fit(X_train, y_train)

print("Models retrained successfully (no leakage).")
```


Cell 15: Evaluate Corrected Models



```python

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report, confusion_matrix

def evaluate_model(model, X_test, y_test, model_name):
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:, 1]
    
    print(f"\n===== {model_name} Results =====")
    print(f"Accuracy : {accuracy_score(y_test, y_pred):.4f}")
    print(f"Precision: {precision_score(y_test, y_pred):.4f}")
    print(f"Recall   : {recall_score(y_test, y_pred):.4f}")
    print(f"F1-Score : {f1_score(y_test, y_pred):.4f}")
    print(f"AUC      : {roc_auc_score(y_test, y_proba):.4f}")
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))

evaluate_model(rf, X_test, y_test, "Random Forest (Corrected)")
evaluate_model(xgb, X_test, y_test, "XGBoost (Corrected)")
```


Cell 16: Improved Models with Class Weighting


```python
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# Optional: Apply SMOTE on training data only
smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print("Before SMOTE:", y_train.value_counts().to_dict())
print("After SMOTE :", pd.Series(y_train_res).value_counts().to_dict())

# Improved Random Forest
rf_imp = RandomForestClassifier(
    n_estimators=400,
    max_depth=10,
    min_samples_split=10,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)
rf_imp.fit(X_train_res, y_train_res)

# Improved XGBoost
xgb_imp = XGBClassifier(
    n_estimators=400,
    max_depth=5,
    learning_rate=0.07,
    scale_pos_weight=2.1,   # roughly majority/minority ratio
    random_state=42,
    eval_metric='logloss'
)
xgb_imp.fit(X_train_res, y_train_res)

print("Improved models trained.")
```


Cell 17: Evaluate Improved Models


```python
evaluate_model(rf_imp, X_test, y_test, "Improved Random Forest")
evaluate_model(xgb_imp, X_test, y_test, "Improved XGBoost")
```


Cell 18: SHAP Interpretability


```python
# ============================================
# Fixed SHAP Diagnostic + Importance Ranking
# ============================================

import shap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

explainer = shap.TreeExplainer(rf_imp)
shap_values = explainer.shap_values(X_test)

print("Type of shap_values:", type(shap_values))
if isinstance(shap_values, list):
    print("Length of list:", len(shap_values))
    print("Shape of shap_values[0]:", np.array(shap_values[0]).shape)
    print("Shape of shap_values[1]:", np.array(shap_values[1]).shape)
    sv = np.array(shap_values[1])
elif isinstance(shap_values, np.ndarray):
    print("Shape of shap_values:", shap_values.shape)
    if shap_values.ndim == 3:
        # shape (n_samples, n_features, n_classes)
        sv = shap_values[:, :, 1]
    else:
        sv = shap_values
else:
    sv = shap_values

print("\nFinal sv shape used for plotting:", sv.shape)
print("X_test shape:", X_test.shape)

# Mean absolute SHAP importance
mean_abs_shap = pd.Series(np.abs(sv).mean(axis=0), index=X_test.columns)
print("\n===== Feature Importance (Mean |SHAP|) =====")
print(mean_abs_shap.sort_values(ascending=False).round(4))

# Summary plot
plt.figure(figsize=(10, 6))
shap.summary_plot(sv, X_test, show=False)
plt.title("SHAP Summary Plot - Improved Random Forest")
plt.tight_layout()
plt.show()
```
